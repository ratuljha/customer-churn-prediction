```python
import sys, os
sys.path.append(os.path.abspath('../src'))
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

sns.set_theme(style='whitegrid')
print('Dependencies loaded successfully!')
```

## 1. Load Preprocessed Data and Trained XGBoost Model

```python
model = joblib.load('../model/churn_model.pkl')
X_test, y_test = joblib.load('../data/test_data.pkl')
feature_names = joblib.load('../model/feature_names.pkl')
print(f'Model loaded: {type(model).__name__}')
print(f'Test samples: {len(X_test):,} | Features: {len(feature_names)}')
```

## 2. Evaluate Model Performance
Checking classification metrics and ROC-AUC score on the test set.

```python
from sklearn.metrics import classification_report, roc_auc_score
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]
print(f'Test ROC-AUC: {roc_auc_score(y_test, y_prob):.4f}')
print('\nClassification Report:')
print(classification_report(y_test, y_pred, target_names=['Retained', 'Churned']))
```

## 3. SHAP Feature Attribution
Interpreting which features contribute most to predicting customer churn risk.

```python
import shap
explainer = shap.TreeExplainer(model)
shap_values = explainer(X_test[:200])
shap.plots.beeswarm(shap_values, max_display=10)
```
